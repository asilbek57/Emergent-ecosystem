```python
import random
import pygame
import sys

cells = []
class Cell:
    def __init__(self, x, y, energy, age=0, is_alive = True):
        if not isinstance(energy, int) or not isinstance(age, int) or not isinstance(is_alive, bool) or not isinstance(x, int) or not isinstance(y, int):
            raise ValueError("Invalid inputs for the Cell attributes")
        self.x = x
        self.y = y
        self.energy = energy
        self.age = age
        self.is_alive = is_alive
    def move(self, dx, dy):
        if not isinstance (dx, int) or not isinstance(dy, int):
            raise ValueError("Coordinates must be integers")

    def update(self):
        if self.is_alive:
            self.age += 1
            self.energy -= 1
            if self.energy <= 0:
                self.is_alive = False
            elif self.age > 100:
                self.is_alive = False
            return f"Cell at ({self.x}, {self.y}) has aged to {self.age} and has {self.energy} energy left."
        else:
            return f"Cell at ({self.x}, {self.y}) is dead."

    def change_color(self):
        max_energy = 10
        ratio = max(0, min(1, self.energy / max_energy)) #this ensures that the ratio does not exceed 1 (which could cause errors with rgb colouring)
        r = 0
        g = int(ratio * 255)
        b = 0
        return (r, g, b)

    def draw(self, screen, cell_size):
        if self.is_alive:
            pixel_x = self.x * cell_size
            pixel_y = self.y * cell_size
            color = self.change_color()
            pygame.draw.rect(screen, color, (pixel_x, pixel_y, cell_size, cell_size))

    def __str__(self):
        return f"Cell({self.x},{self.y}, {self.energy}, {self.age}, {self.is_alive})"

class Population:
    def __init__(self):
        self.cells = []

    def add_cell(self, cell):
        self.cells.append(cell)

    def new_cell(self, x, y, energy):
        new_cell = Cell(x,y, energy)
        self.cells.append(new_cell)

    def random_population(self, num_cells):
        for cell in range(num_cells):
            x = random.randint(0,49)
            y = random.randint(0,49)
            cell = Cell(x, y, 10)
            self.cells.append(cell)

    def update_all(self):
        for cell in self.cells:
            cell.update()

        self.cells = [cell for cell in self.cells if cell.is_alive]

        for cell in self.cells:
            print(f"Cell at ({cell.x}, {cell.y}) has aged to {cell.age} and has {cell.energy} energy left")
        
        return f"Population : {len(self.cells)} cell(s) alive"

pop = Population()


def run_game():
    pygame.init()
    CELL_SIZE = 16
    GRID_WIDTH = 50
    GRID_HEIGHT = 50
    WINDOW_WIDTH = GRID_WIDTH * CELL_SIZE
    WINDOW_HEIGHT = GRID_HEIGHT * CELL_SIZE

    screen = pygame.display.set_mode((WINDOW_WIDTH, WINDOW_HEIGHT))
    pygame.display.set_caption("Emergent Ecosystem")

    clock = pygame.time.Clock()
    FPS = 1

    
    pop.random_population(50)
    running = True
    while running:
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
        
        pop.update_all()

        screen.fill((30, 30, 30)) #dark background 
        for cell in pop.cells:
                    cell.draw(screen, CELL_SIZE)

        pygame.display.flip() #always updating the display
        clock.tick(FPS)
    pygame.quit()
    sys.exit()


run_game()
```
